```python
age_list = [24, 25, 26, 27, 28]

print(age_list)
```

```python
# Create a list of five names
name_list = ["Arun", "Bala", "Charan", "Divya", "Esha"]

# a. Append Yazhini
name_list.append("Yazhini")
print("After append:", name_list)

# b. Insert 30 at index 2
age_list.insert(2, 30)
print("After insert:", age_list)

# c. Remove Yazhini
name_list.remove("Yazhini")
print("After remove:", name_list)

# d. Pop the last element
age_list.pop()
print("After pop:", age_list)

# e. Extend age_list
age_list.extend([29, 30, 26])
print("After extend:", age_list)

# f. Sort in descending order
age_list.sort(reverse=True)
print("Descending order:", age_list)

# g. Find maximum, minimum and sum
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python
# a. First element
print("First element:", name_list[0])

# b. Last element
print("Last element:", name_list[-1])

# c. Elements from index 2 to 4
print("Index 2 to 4:", name_list[2:5])

# d. Reverse order
print("Reverse order:", name_list[::-1])
```

```python
# a. Create a dictionary of five students
student_marks = {
    "Arun": 75,
    "Bala": 85,
    "Charan": 90,
    "Divya": 78,
    "Esha": 88
}

# b. Access and print a student's mark
print("Arun's mark:", student_marks["Arun"])

# c. Add Janani with mark 80
student_marks["Janani"] = 80
print("After adding Janani:", student_marks)

# d. Update an older student's mark to 82
student_marks["Bala"] = 82
print("After updating Bala:", student_marks)

# e. Print keys, values and items
print("Student names:", student_marks.keys())
print("Student marks:", student_marks.values())
print("Key-value pairs:", student_marks.items())
```

```python
# a. Create a set
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print("My set:", my_set)

# Explanation of output
print("Duplicate values are automatically removed.")

# b. Attempt to change a set value
try:
    my_set[4] = 's'
except TypeError:
    print("Error: Sets do not support item assignment.")

# c. Create two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# d. Find union and intersection
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```
